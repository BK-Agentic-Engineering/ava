# LifeMate — official timing interface

Live audio, existing smolagents + llama.cpp pipeline. Full private split by default. No evaluation or replay.


```python
from lifemate.deploy import set_seed
set_seed()
```

```python
from pathlib import Path
from lifemate.deploy import load_models
ROOT = Path("/data")
SPLIT = "private_test"
OUTPUT = Path("/result")
models = load_models(Path("/app/models"), OUTPUT / "jupyter_runtime")
asr, backend = models.__enter__()
```

```python
from lifemate.deploy import run_split
try:
    errors = run_split(ROOT, SPLIT, OUTPUT, asr, backend, submission_name="jupyter_submission.jsonl")
finally:
    models.__exit__(None, None, None)
if errors:
    raise RuntimeError(f"{errors} samples had execution errors; inspect /result/jupyter_submission/results.jsonl")
```
